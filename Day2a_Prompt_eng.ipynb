Step 1: Why Big-O Exists
Suppose you write two programs.
Program A searches a number in an array of 10 elements.
Program B searches in an array of 10 lakh elements.
Obviously, execution time changes when input grows.
Big-O helps us answer:
	Will the algorithm still be fast for huge inputs?
	How does performance scale?
	Which solution is better in interviews?
________________________________________
Step 2: Think in Terms of Input Size
We usually represent input size as n.
Examples:
	Array size = n
	Number of students = n
	Number of nodes in graph = n
Now we study how operations increase when n increases.
________________________________________
Step 3: Most Important Complexities for Placements
These are the ones you MUST know for coding rounds.
________________________________________
1. O(1) — Constant Time
Time does NOT increase with input size.
Example:
int x = arr[0];
Whether array has 10 elements or 10 crore elements,
accessing first element takes same time.
Interview line:
“Execution time remains constant.”
________________________________________
2. O(log n) — Logarithmic Time
Very efficient.
Input gets reduced repeatedly.
Main example:
	Binary Search
Each step cuts array into half.
Example:
	1000 elements
	Step 1 → 500
	Step 2 → 250
	Step 3 → 125
So operations grow very slowly.
Interviewers LOVE when you identify binary search as O(log n).
________________________________________
3. O(n) — Linear Time
You traverse entire input once.
Example:
for(int i=0;i<n;i++)
If n doubles, work doubles.
Common examples:
	Finding max element
	Linear search
	Counting frequency
________________________________________
4. O(n log n)
Usually seen in efficient sorting algorithms.
Examples:
	Merge Sort
	Heap Sort
	Quick Sort (average case)
This is considered GOOD for sorting.
________________________________________
5. O(n²) — Quadratic Time
Nested loops.
Example:
for(i=0;i<n;i++)
{
   for(j=0;j<n;j++)
}
If n becomes 1000:
operations ≈ 10 lakh.
Very slow for large inputs.
Common in:
	Bubble Sort
	Selection Sort
	Comparing every pair
________________________________________
6. O(2ⁿ) — Exponential
Danger zone.
Usually recursion generating all possibilities.
Example:
	Fibonacci recursion
	Subset generation (brute force)
Very bad for large n.
________________________________________
Step 4: Order of Complexities (Very Important)
From BEST to WORST:
O(1)<O(log⁡n)<O(n)<O(nlog⁡n)<O(n^2)<O(2^n)

This ordering is extremely common in MCQs and interviews.
________________________________________
Step 5: How to Calculate Big-O Quickly
This is where students struggle.
Let’s simplify.
________________________________________
Rule 1: Ignore Constants
Example:
for(i=0;i<2*n;i++)
Complexity is:
O(n)
Not O(2n).
Because Big-O focuses on growth trend.
________________________________________
Rule 2: Drop Smaller Terms
Example:
O(n² + n + 5)
Answer:
O(n²)
Because n² dominates when n becomes huge.
________________________________________
Rule 3: Sequential Loops Add
for(i=0;i<n;i++)   // O(n)

for(j=0;j<n;j++)   // O(n)
Total:
O(n + n) = O(n)
________________________________________
Rule 4: Nested Loops Multiply
for(i=0;i<n;i++)
{
   for(j=0;j<n;j++)
}
Complexity:
O(n × n) = O(n²)
________________________________________
Step 6: Time Complexity vs Space Complexity
Students often mix these.
Time Complexity
How much time algorithm takes.
Space Complexity
How much extra memory algorithm uses.
Example:
	Merge Sort → O(n) extra space
	Bubble Sort → O(1) extra space
Interviewers may ask both.

1. Time Complexity
Time complexity tells you how the execution time of an algorithm grows when input size increases. In placement interviews, after writing code, interviewers often ask, “What is the time complexity?” A single loop is usually O(n), nested loops often become O(n²), and binary search is O(log n). For TCS, Infosys, and Cognizant, you should be comfortable identifying complexity directly from loops and operations.
2. Space Complexity
Space complexity measures how much extra memory an algorithm uses while running. Students usually focus only on execution time and forget memory usage. For example, using an extra array of size n gives O(n) space, while using only a few variables gives O(1) space. In interviews, sometimes a faster algorithm uses more memory, and you may be asked to discuss this trade-off.
3. Worst-Case, Average-Case, and Best-Case Analysis
Algorithms do not always behave the same for every input. Best case means minimum work, worst case means maximum work, and average case is the expected behavior. For example, in linear search, finding the element at the first position is best case O(1), while checking the entire array is worst case O(n). Interviewers usually prefer discussing the worst-case complexity.
4. Common Complexity Classes (O(1), O(log n), O(n), O(n²), etc.)
You must remember common Big-O classes because they appear frequently in coding rounds. O(1) means constant time like accessing an array element, O(log n) appears in binary search, O(n) in single traversal problems, and O(n²) in nested loops. The important idea is that lower growth rates scale better when input size becomes very large.
5. Complexity Analysis of Loops and Nested Loops
Most placement questions can be solved by understanding loops properly. A loop running n times gives O(n). Two nested loops each running n times give O(n²). If one loop reduces input by half every iteration, complexity becomes O(log n). Instead of counting code lines, always count how many times operations execute with respect to input size—this is the key interview skill.
Big-O notation helps you understand how an algorithm behaves as input size grows. Focus on time complexity (execution speed) and space complexity (memory used). Also learn best, average, and worst cases, though interviews mostly discuss worst case. Remember common classes like O(1), O(log n), O(n), and O(n²). Finally, practice loop analysis because single and nested loops decide complexity in many coding questions.
